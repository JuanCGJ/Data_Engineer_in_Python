## Chapter 1 — Built-in functions, modules and packages

### 1.1 Built-in functions
Built-in functions solve common tasks with less code: `max`, `min`, `sum`, `round`, `len`, `sorted` (plus `print`, `type`, `help`). Compared to a manual loop they are shorter, reusable and less error-prone.

```python
preparation_times = [19.23, 15.67, 48.57, 23.45, 12.06, 34.56, 45.67]

print("Longest:", max(preparation_times), "| Shortest:", min(preparation_times))
print("Total:", round(sum(preparation_times), 1))
print("Orders:", len(preparation_times), "| Characters in 'Burger Hub':", len("Burger Hub"))
print("Average:", sum(preparation_times) / len(preparation_times))
print(sorted(preparation_times))
print(sorted("George"))
```

```python
# The same total with a manual loop: more code for the same result
time_count = 0
for t in preparation_times:
    time_count += t
print(round(time_count, 2) == round(sum(preparation_times), 2))
```

### 1.2 Modules
A module is a `.py` file with functions and attributes we can import instead of rewriting code. Python ships with ~200 built-in modules. **Functions** perform tasks (use parentheses); **attributes** return values (no parentheses).

- `os` interacts with the operating system (current directory, files, environment variables).
- `string` provides ready-made character sets for validating text.

```python
import os
import string

print(type(os))
work_dir = os.getcwd()                      # function
print("Current folder:", os.path.basename(work_dir))
print("Files here:", sorted(os.listdir("."))[:5])
print("PATH defined in environment:", "PATH" in os.environ)   # attribute (a mapping)

print(string.digits)
print(string.ascii_lowercase)
print(string.punctuation)
```

```python
# Practical use: validate a password with the string module
def check_password(pwd):
    return {
        "has_digit": any(ch in string.digits for ch in pwd),
        "has_lower": any(ch in string.ascii_lowercase for ch in pwd),
        "has_symbol": any(ch in string.punctuation for ch in pwd),
    }

print(check_password("burger2024!"))
print(check_password("burger"))
```

We import it, usually with an alias. `pandas` is the standard for tabular data. ## Chapter 2 — Custom functions

### 2.1 Defining a function
Following **DRY (Don't Repeat Yourself)**, a block becomes a function when it is long, complex or used often. `def name(arguments):` defines it, `return` sends the output back, and variables created inside only exist inside the function.

```python
def average(values):
    average_value = sum(values) / len(values)
    rounded_average = round(average_value, 2)
    return rounded_average

orders = [12, 8, 10, 9, 15, 21, 16]
print(average(preparation_times))
print(average(orders))

average_time = average(preparation_times)   # store the output
```

### 2.2 Positional, keyword and default arguments
- **Positional**: passed in order.
- **Keyword**: passed as `name=value`, which is clearer to read.
- **Default**: a value used when the caller doesn't provide one (`help()` shows them, e.g. `round(number, ndigits=None)`).

```python
print(round(3.1415926535, 2))                      # positional
print(round(number=3.1415926535, ndigits=2))       # keyword

def average(values, rounded=False):
    average_value = sum(values) / len(values)
    if rounded:
        return round(average_value, 2)
    return average_value

print(average(preparation_times))                          # uses the default
print(average(values=preparation_times, rounded=True))     # overrides it
```

### 2.3 Docstrings
A docstring is the string right after `def` that explains what a function does, its arguments and its return value. It is stored in `__doc__` and shown by `help()`.

```python
def average(values, rounded=False):
    """
    Find the mean in a sequence of values, optionally rounded to two decimal places.

    Args:
        values (list): A list of numeric values.
        rounded (bool): Round the result to two decimals if True.

    Returns:
        float: The mean of values.
    """
    average_value = sum(values) / len(values)
    return round(average_value, 2) if rounded else average_value

print(average.__doc__)
```

### 2.4 Arbitrary arguments: `*args` and `**kwargs`
- `*args` accepts any number of positional arguments and packs them into a **tuple**.
- `**kwargs` accepts any number of keyword arguments and packs them into a **dictionary**.
- `*` and `**` also **unpack** lists and dictionaries when calling a function.

```python
def average_args(*args):
    return round(sum(args) / len(args), 2)

print(average_args(15, 29, 4, 13, 11, 8))
print(average_args(*[15, 29], *[4, 13], *[11, 8]))   # unpack several lists

def average_kwargs(**kwargs):
    return round(sum(kwargs.values()) / len(kwargs), 2)

print(average_kwargs(a=15, b=29, c=4, d=13, e=11, f=8))
print(average_kwargs(**{"a": 15, "b": 29}, **{"c": 4, "d": 13}, **{"e": 11, "f": 8}))
```

## Chapter 3 — Lambda functions and error handling

### 3.1 Lambda functions
`lambda arguments: expression` creates a small anonymous function with no `return` statement. It is ideal for simple, one-off tasks, often inside `map()`, which applies a function to every element of an iterable. Complex or repeated logic should stay in a regular `def` function.

```python
average = lambda x: sum(x) / len(x)
print(average([3, 6, 9]))
print((lambda x: sum(x) / len(x))([3, 6, 9]))     # call it directly

power = lambda x, y: x ** y
print(power(2, 3))

names = ["john", "sally", "leah"]
capitalized = map(lambda x: x.capitalize(), names)
print(capitalized)            # a lazy map object
print(list(capitalized))
```

### 3.2 Errors (exceptions)
An error is code that breaks a rule, and it stops the program. Two common ones: **TypeError** (incompatible types) and **ValueError** (right type, unacceptable value). The traceback shows where it happened, including inside package source code such as pandas.

```python
for bad_code in ['"Hello" + 5', 'float("Hello")']:
    try:
        eval(bad_code)
    except Exception as e:
        print(f"{bad_code:<16} -> {type(e).__name__}: {e}")

print(float("2"))     # a valid value converts fine

products = pd.DataFrame({"ID": ["ABC1"], "price": [29.99]})
try:
    products["tag"]
except KeyError as e:
    print("KeyError from pandas source code:", e)
```

### 3.3 Handling errors: `try-except` vs. `raise`
Design thinking: anticipate how people might misuse our function (wrong data type, wrong number of arguments) and decide what should happen.

- `try-except` **catches** the error, shows a helpful message and lets the rest of the program keep running.
- `raise` **produces** a clear, specific error and stops execution, so bad data never moves forward.

```python
sales_dict = {"cust_id": ["JL93", "MT12", "IY64"], "order_value": [43.21, 68.70, 82.19]}

def average_safe(values):
    """try-except: return a friendly message instead of crashing."""
    try:
        return round(sum(values) / len(values), 2)
    except TypeError:
        print("average() accepts a list or set. Please provide a correct data type.")

def average_strict(values):
    """raise: refuse invalid input explicitly."""
    if type(values) in (list, set):
        return round(sum(values) / len(values), 2)
    raise TypeError("average() accepts a list or set, please provide a correct data type.")

average_safe(sales_dict)              # prints a message, program continues
print(average_strict([43.21, 68.70, 82.19]))

try:
    average_strict(sales_dict)
except TypeError as e:
    print("Raised:", e)
```

## 🎯 Key takeaways

- Reach for built-ins, modules (`os`, `string`) and packages (`pandas`) before writing code from scratch.
- Good functions are small, documented (docstrings) and flexible (defaults, `*args`, `**kwargs`).
- Lambdas are for short, one-off transformations; `def` for anything reused or complex.
- `try-except` keeps a program running; `raise` stops bad data early with a clear message.

**In a Data Analyst / Data Engineer role:** pipelines are built from small, tested, documented functions. `os` handles file paths and environment configuration, `pandas` loads and inspects data, and explicit error handling (`raise` on bad input, `try-except` around unreliable sources) is what makes an ETL job fail loudly instead of loading corrupt data.
