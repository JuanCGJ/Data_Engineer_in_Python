## Chapter 1 — Python basics

### 1.1 `print()` and comments
`print()` is a built-in function that displays results, which is the simplest way to test code. Comments start with `#` and document the code without affecting it.

```python
# This is a comment: Python ignores it
print("Hello, world!")
print('Single quotes work too')
```

### 1.2 Variables and data types
Variables store values so we can reuse and update them. Names use lowercase and underscores. Python assigns the data type automatically, and `type()` tells us which one it is, which in turn tells us what operations are allowed.

```python
ingredient_name = "Tomatoes"        # str
ingredient_quantity = 2             # int
new_ingredient_quantity = 1.5       # float
is_in_stock = True                  # bool (capitalized, no quotes)

for value in [ingredient_name, ingredient_quantity, new_ingredient_quantity, is_in_stock]:
    print(repr(value), "->", type(value))

# Variables can be updated
ingredient_quantity = 1
print("Updated quantity:", ingredient_quantity)
```

### 1.3 Operators
Arithmetic operators work like regular math on numbers. On strings, only `+` (concatenate) and `*` (repeat) work.

```python
print(2 + 1.5, 10 - 3, 4 * 2, 7 / 2)
print("Hi" + "There")
print("Hi" * 3)
```

## Chapter 2 — Data structures

### 2.1 Strings and string methods
Single or double quotes both work; double quotes avoid errors when the text contains an apostrophe. Triple quotes create multi-line strings (instructions, documentation). A **method** is a function that belongs to a specific data type, called as `value.method()`.

```python
special = "Chef's special seasoning"   # apostrophe inside double quotes is fine

recipe_instructions = """1. Boil salted water and cook the pasta
2. Saute garlic in olive oil
3. Add tomatoes and simmer for 10 minutes"""
print(recipe_instructions)

welcome = "Welcome to the recipe scaler, George".replace("George", "John")
print(welcome)
print("Basil Leaves".lower(), "|", "Basil Leaves".upper())   # standardize user input
```

### 2.2 Lists
A list stores many values in one variable. It is **ordered and zero-indexed**: negative indexes count from the end, and slicing `[start:stop:step]` returns everything up to, but not including, `stop`.

```python
ingredients = ["pasta", "tomatoes", "garlic", "basil", "olive oil", "salt"]

print(type(ingredients))
print(ingredients[0], "|", ingredients[3], "|", ingredients[-1])  # first, fourth, last
print(ingredients[1:3])    # 2nd and 3rd
print(ingredients[3:])     # from index 3 onwards
print(ingredients[:3])     # first three
print(ingredients[::2])    # every second element
print(ingredients[1::3])   # every third, starting at index 1
```

### 2.3 Dictionaries
Dictionaries store **key-value pairs**, which is exactly what a recipe scaler needs: link each ingredient to its quantity. We access values by key, add or update with assignment, and if a key is repeated only the last value is kept.

```python
recipe = {"pasta": 500, "tomatoes": 400, "garlic": 15, "basil": 20, "olive oil": 30, "salt": 5}

print(recipe["pasta"])
print(recipe.keys())
print(recipe.values())
print(list(recipe.items())[:2])

recipe["parmesan"] = 50      # add a new pair
recipe["pasta"] = 1000       # update an existing key
print(recipe)

duplicate = {"garlic": 5, "garlic": 15}
print("Duplicate key keeps the last value:", duplicate["garlic"])
```

### 2.4 Sets and tuples
- **Set**: unique values, no index, fast membership checks. Ideal to remove duplicates. `sorted()` returns a list.
- **Tuple**: ordered and **immutable** (no adding, removing or changing). Good for fixed identifiers or options.

```python
ingredients_list = ["pasta", "tomatoes", "garlic", "basil", "olive oil", "pasta", "salt"]
unique_ingredients = set(ingredients_list)
print(type(unique_ingredients), len(ingredients_list), "->", len(unique_ingredients), "items")
print(sorted(unique_ingredients))

try:
    unique_ingredients[0]
except TypeError as e:
    print("TypeError:", e)          # sets cannot be subset with []

serving_sizes = (1, 2, 4, 6, 8)
print(serving_sizes[1])
try:
    serving_sizes[0] = 3
except TypeError as e:
    print("TypeError:", e)          # tuples are immutable
```

## Chapter 3 — Control flow

### 3.1 Comparison operators and conditionals
Comparisons (`==`, `!=`, `<`, `>`, `<=`, `>=`) return booleans; strings compare alphabetically. `if` / `elif` / `else` run code only when a condition is met. The block after the colon **must be indented**.

```python
print(2 == 3, 2 != 3, 5 <= 7, "James" > "Brian")

required_quantity = 500
pasta_quantity = 200

if pasta_quantity >= required_quantity:
    print("You have enough pasta!")
elif pasta_quantity >= 300:
    print("Nearly enough pasta. Try a smaller portion.")
else:
    print("Not enough pasta.")
```

### 3.2 `for` loops
A `for` loop runs an action for every value in an iterable (list, string, dictionary, `range`). Combined with conditionals it replaces many individual comparisons.

```python
quantities = [1000, 800, 40, 30, 30, 15]
for qty in quantities:
    if qty > 500:
        print(qty, "Plenty in stock")
    elif qty >= 100:
        print(qty, "Enough for a small portion")
    else:
        print(qty, "Nearly out!")
```

```python
# Looping through a string, a dictionary and a range
print([letter for letter in "pasta"])

ingredients = {"pasta": 500, "tomatoes": 400, "garlic": 30}
factor = 2
for item, qty in ingredients.items():
    print(item, ":", qty * factor, "grams")

for i in range(1, 6):     # stop is excluded -> 1..5
    print(i, end=" ")
```

### 3.3 `while` loops
A `while` loop repeats **while** a condition stays true. The variable in the condition must change inside the loop, otherwise it never ends (infinite loop). `break` exits a loop early.

```python
ingredients_to_add = 5
items_added = 0

while items_added < ingredients_to_add:
    items_added += 1                       # without this line the loop never ends
    remaining = ingredients_to_add - items_added
    if remaining > 3:
        print("Several ingredients remaining")
    elif remaining >= 1:
        print("Almost done!")
    else:
        print("Shopping list complete!")
```

```python
# break: stop as soon as we find what we need
for ingredient in ["pasta", "garlic", "salt", "basil"]:
    if ingredient == "salt":
        print("Found salt, stopping the search")
        break
```

### 3.4 Building a workflow: `in`, `not`, `and`, `or`, `+=`, `.append()`
These keywords let us combine loops and conditions into real logic: membership checks, multiple requirements, counters and collecting results in a list.

```python
recipe = {"pasta": 500, "tomatoes": 400, "garlic": 15, "basil": 20, "salt": 5}
pantry = {"pasta": 600, "garlic": 30, "basil": 10}

print("pasta" in recipe)                       # in
print("salt" not in pantry)                    # not
print(pantry["pasta"] >= 500 and pantry["garlic"] >= 15)   # and
print(pantry["basil"] >= 20 or pantry["garlic"] >= 15)     # or

shopping_list = []
checked = 0
for ingredient, qty_needed in recipe.items():
    checked += 1
    if ingredient not in pantry or pantry[ingredient] < qty_needed:
        shopping_list.append(ingredient)

print(f"Checked {checked} ingredients -> buy: {shopping_list}")
```

## Putting it together: a mini recipe scaler

The course project in a few lines: scale a recipe to a number of servings and flag what must be bought.

```python
base_servings = 4
recipe = {"pasta": 500, "tomatoes": 400, "garlic": 15, "basil": 20, "olive oil": 30, "salt": 5}
pantry = {"pasta": 1000, "tomatoes": 300, "garlic": 40, "salt": 100}

target_servings = 6
factor = target_servings / base_servings

to_buy = []
for item, qty in recipe.items():
    needed = round(qty * factor, 1)
    have = pantry.get(item, 0)
    status = "OK" if have >= needed else "BUY"
    if status == "BUY":
        to_buy.append(item)
    print(f"{item:<10} need {needed:>6} g | have {have:>5} g | {status}")

print("\nShopping list:", to_buy)
```

## 🎯 Key takeaways

- Python's data types define which operations are valid; `type()` is the quickest sanity check.
- Choosing the right structure matters: **lists** for ordered sequences, **dictionaries** for lookups by key, **sets** for uniqueness, **tuples** for fixed values.
- Conditionals plus loops are the core of any data validation step.
- Developer habits (small steps, testing with `print`, comments) make code easier to maintain.

**In a Data Analyst / Data Engineer role:** these are the building blocks of every pipeline: dictionaries map to JSON records and config files, sets deduplicate keys, and loops with conditions are how raw rows get validated, filtered and routed before loading.
